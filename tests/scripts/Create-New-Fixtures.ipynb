# Test Fixture Generation

This notebook creates test fixtures from real MKV recordings for the Minutia pipeline
test suite.

## What are Test Fixtures?

Test fixtures are small, controlled samples of real data used for automated testing.
They provide:

- **Reproducible test data** - Same inputs produce same outputs across test runs
- **Fast test execution** - Small files (5-30 seconds) for quick feedback
- **Diverse scenarios** - Different audio characteristics to test edge cases
- **Regression protection** - Detect when changes break existing functionality

## Our Test Strategy

The fixtures support a 3-tier testing approach:

1. **Unit tests** (< 1s) - Mock dependencies, test individual functions with real audio
files
2. **Integration tests** (< 30s) - Real Whisper tiny model, real FFmpeg, mocked APIs
3. **End-to-end tests** (full pipeline) - Complete MKV → summary flow for regression
testing

## Generated Artifacts

For each fixture, we create:
- **Video clip** (.mkv) - Short segment from original recording
- **Audio tracks** (.wav) - Extracted microphone/desktop audio
- **Metadata** (.json) - Properties, source info, and validation data

These fixtures will be processed through the full pipeline to create "golden samples" -
known-good reference outputs for regression testing.
